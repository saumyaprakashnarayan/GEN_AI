```python
from langchain_chroma import Chroma
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import OllamaEmbeddings
```

```python
loader=TextLoader("speech.txt")
data=loader.load()n
```

```python
splitter=RecursiveCharacterTextSplitter(chunk_size=200,chunk_overlap=30)
splits=splitter.split_documents(data)
```

```python
embeddings=OllamaEmbeddings()
vector_db=Chroma.from_documents(documents=splits,embedding=embeddings)
```

```python
query="what does the speaker is about"
query_result=vector_db.similarity_search()
```

```python
#saving to the disk
vector_db=Chroma.from_documents(documents=splits,embedding=embeddings,persist_directory="./chroma_db")# embedding and storing in vectordb
```

```python
#loading from local space

db2=Chroma(persist_directory="./chroma_db",embedding_function=embeddings)
docs=db2.similarity_search(query)
```

```python
docs2=db2.similarity_search_with_score(query)
```
