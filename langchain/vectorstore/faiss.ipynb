```python
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import FAISS
```

```python
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import CharacterTextSplitter
```

```python
loader=TextLoader("speech.txt")
documents=loader.load()

text_splitter=CharacterTextSplitter(chunk_size=100,chunk_overlap=30)
docs=text_splitter.split_documents(documents)
```

```python
embeddings=OllamaEmbeddings()
```

```python
db=FAISS.from_documents(docs,embeddings) # perform embeddings and store the vectors
```

```python
#query
query="how does the speaker describe about the war?"
query_result=db.similarity_search(query)

query_result[0].page_content 
```

```python
#RETRIEVER
retriever=db.as_retriever()
retriever.invoke(query)
```

```python
query_vector=embeddings.embed_query(query)#to convert the query into vectors
query_vector_result=db.similarity_search_by_vector(query_vector)
```

```python
#loading and saving
db.save_local("faiss_index")#saving locally
new_db=FAISS.load_local("faiss_index",embeddings,allow_dangerous_deserialization=True)
```

```python
docs=new_db.similarity_search(query)
```
