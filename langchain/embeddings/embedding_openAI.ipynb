```python
embedding1024=OpenAIEmbeddings(model="text-embedding-3-large",dimensions=1024)
```

```python
#text loader
from langchain_community.document_loaders import TextLoader 
loader=TextLoader("speech.txt")

text_documents=loader.load() #to create langchain document object
```

```python
#vector embedding and vector storeDB
from langchain_community.vectorstores import Chroma
db=Chroma.from_documents(text_documents,embedding1024)
```

```python
#retrive the query and perfome similarity search on the vector databse
query="t will be all the easier for us to conduct ourselves as belligerents in "
retrieved_result= db.similarity_search(query)#performing similarity search
```
